```python
# build embeddings
from sentence_transformers import SentenceTransformer
```

```python
model = SentenceTransformer("all-MiniLM-L6-v2")
```

```python
docs = [
    "I love luxeries cars, such as Mercedes BenZ",
    "Luxeries cars are very comfortable",
    "Luxeries cars don't only appear beautiful, but also it should be good in mechanical wise"
]
```

```python
embeddings = model.encode(docs)
```

```python
# Query
query = "Mercedes"
query_embedding = model.encode(query)
```

```python
from sklearn.metrics.pairwise import cosine_similarity
score = cosine_similarity([query_embedding], embeddings)
score
```

```python
import numpy as np
best_match = np.argmax(score)
```

```python
docs[best_match]
```

```python
import chromadb
client = chromadb.Client()
collection = client.create_collection("Route")
collection  # Collection(name=Route)
```

```python
collection.add(
    documents=[
        "Hend Loves Cars",
        "Cars are comfortable and beautiful"
    ], ids=["1", "2"]
)
```

```python
out = collection.query(query_texts=["who loves cars?"], n_results=1)
```
